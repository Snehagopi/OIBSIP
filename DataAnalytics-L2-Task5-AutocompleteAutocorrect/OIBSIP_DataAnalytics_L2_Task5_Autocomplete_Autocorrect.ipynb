```python
from nltk.corpus import gutenberg

text = gutenberg.raw("austen-emma.txt")

print("Text loaded successfully!")
print("Number of characters:", len(text))
print("\nFirst 500 characters:\n")
print(text[:500])
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text.lower())

print("Total tokens:", len(tokens))
print("\nFirst 30 tokens:")
print(tokens[:30])
```

```python
import string

words = [
    word for word in tokens
    if word.isalpha()
]

print("Total words:", len(words))
print("\nFirst 30 words:")
print(words[:30])
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

clean_words = [
    word for word in words
    if word not in stop_words
]

print("Words after stopword removal:", len(clean_words))
print("\nFirst 30 clean words:")
print(clean_words[:30])
```

```python
from nltk.util import bigrams

bigram_list = list(bigrams(clean_words))

print("Total bigrams:", len(bigram_list))
print("\nFirst 20 bigrams:")
print(bigram_list[:20])
```

```python
from nltk.util import trigrams

trigram_list = list(trigrams(clean_words))

print("Total trigrams:", len(trigram_list))
print("\nFirst 20 trigrams:")
print(trigram_list[:20])
```

```python
from collections import Counter

bigram_counts = Counter(bigram_list)

print("Top 20 bigrams:")
print(bigram_counts.most_common(20))
```

```python
trigram_counts = Counter(trigram_list)

print("Top 20 trigrams:")
print(trigram_counts.most_common(20))
```

```python
from collections import defaultdict

bigram_next_words = defaultdict(Counter)

for first_word, second_word in bigram_list:
    bigram_next_words[first_word][second_word] += 1

print("Bigram autocomplete model created successfully!")
```

```python
def autocomplete_bigram(prefix, n=3):
    prefix = prefix.lower()
    
    if prefix in bigram_next_words:
        return bigram_next_words[prefix].most_common(n)
    else:
        return []

print(autocomplete_bigram("the"))
```

```python
prefixes = [
    "the",
    "good",
    "every",
    "one",
    "she",
    "you",
    "will",
    "could",
    "when",
    "would"
]

for prefix in prefixes:
    print(prefix, "->", autocomplete_bigram(prefix, 3))
```

```python
trigram_next_words = defaultdict(Counter)

for first_word, second_word, third_word in trigram_list:
    prefix = (first_word, second_word)
    trigram_next_words[prefix][third_word] += 1

def autocomplete_trigram(word1, word2, n=3):
    prefix = (word1.lower(), word2.lower())
    
    if prefix in trigram_next_words:
        return trigram_next_words[prefix].most_common(n)
    else:
        return []

print(autocomplete_trigram("she", "was", 3))
```

```python
vocabulary = set(clean_words)

print("Vocabulary size:", len(vocabulary))
print("First 20 words:", list(vocabulary)[:20])
```

```python
def levenshtein_distance(word1, word2):
    m = len(word1)
    n = len(word2)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost
            )

    return dp[m][n]


print("Distance:", levenshtein_distance("hello", "helo"))
```

```python
def autocorrect(word):
    word = word.lower()

    if word in vocabulary:
        return word

    best_word = None
    best_distance = float("inf")

    for candidate in vocabulary:
        distance = levenshtein_distance(word, candidate)

        if distance < best_distance:
            best_distance = distance
            best_word = candidate

    return best_word


print("Autocorrect:", autocorrect("helo"))
```

## Conclusion

This project applied Natural Language Processing techniques to build autocomplete and autocorrect systems.

Text preprocessing included lowercasing, tokenization, punctuation removal, and stopword removal. Bigram and trigram models were developed to predict the next word, and autocomplete was tested using multiple prefixes.

For autocorrect, Levenshtein edit distance was used to identify possible spelling corrections. The system was tested using deliberately misspelled words and evaluated using Accuracy, Precision, Recall, and a Confusion Matrix.

The project demonstrates how traditional NLP techniques can be used for text prediction and spelling correction. The main limitations are the size and style of the corpus, vocabulary coverage, and the computational cost of comparing words using edit distance.
