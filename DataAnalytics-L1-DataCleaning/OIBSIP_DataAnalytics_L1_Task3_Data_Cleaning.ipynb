```python
def check_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = ((df[column] < lower) | (df[column] > upper)).sum()
    
    print(column)
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Outliers:", count)
    print()

check_outliers('age')
check_outliers('fare')
```

### Outlier Handling

Outliers were detected in numerical columns using the IQR method. Some extreme values, particularly in the Fare column, represent legitimate high-value tickets rather than data-entry errors. Therefore, these outliers were retained to avoid removing valid observations from the dataset.

```python
text_columns = ['sex', 'embarked', 'class', 'who', 'deck', 'embark_town', 'alive']

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

df[text_columns].head()
```

```python
df['survived'] = df['survived'].astype(int)
df['pclass'] = df['pclass'].astype(int)
df['sibsp'] = df['sibsp'].astype(int)
df['parch'] = df['parch'].astype(int)
df['fare'] = df['fare'].astype(float)

df.dtypes
```

### Before vs After Cleaning

The data cleaning process reduced missing values to zero and removed duplicate records where present. Text fields were standardized, numerical data types were corrected, and the cleaned dataset was checked for invalid ranges and outliers. Legitimate outliers were retained rather than removed.
